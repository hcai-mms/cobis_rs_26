```python
import pandas as pd
```

```python
ed = pd.read_csv('experiments/example/input/dataset.inter', sep='\t')
ed
```

```python
# n-core filtering
u = list(ed['user_id:token'].sample(150))
it = list(ed['item_id:token'].sample(450))

sm = ed[ed['user_id:token'].isin(u) & ed['item_id:token'].isin(it)]

n = 5
flt = True

while flt:
    tmp_u = sm.groupby('user_id:token').count()
    tmp_i = sm.groupby('item_id:token').count()
    
    u2drop = tmp_u[tmp_u['item_id:token'] < n].index
    i2drop = tmp_i[tmp_i['user_id:token'] < n].index

    if (len(u2drop) + len(i2drop)) > 0:
        sm = sm[(~sm['user_id:token'].isin(u2drop)) & (~sm['item_id:token'].isin(i2drop))]
    else:
        flt = False

sm
```

```python
sm.to_csv('experiments/debug2/input/dataset_n-core.inter', sep='\t', index=False)
```

```python
# reindexing
sm = pd.read_csv('experiments/debug2/input/dataset.inter', sep='\t')
sm
```

```python
ucol = list(sm['user_id:token'])
icol = list(sm['item_id:token'])

u_uni = list(sm['user_id:token'].unique())
i_uni = list(sm['item_id:token'].unique())

reindexed_u = [u_uni.index(el) for el in ucol]
reindexed_i = [i_uni.index(el) for el in icol]

sm2 = pd.DataFrame(data = {'user_id:token' : reindexed_u, 'item_id:token' : reindexed_i})
sm2
```

```python
sm2.to_csv('experiments/debug2/input/dataset_n-core_reindexed.inter', sep='\t', index=False)
```

```python
# reindexing tracks
idat = pd.read_csv('experiments/debug2/input/tracks.tsv', sep='\t', names=['artist', 'track', 'country'])
ni_dat = idat.iloc[i_uni].sort_index()
ni_dat.index = range(len(i_uni))
ni_dat.to_csv('experiments/debug2/input/tracks_reindexed.tsv', sep='\t', header = 0, index= False)
```

```python
#reindexing users
udat = pd.read_csv('experiments/debug2/input/demographics.tsv', sep='\t', names=['country', 'age', 'gender', 'date'])
nu_dat = udat.iloc[u_uni].sort_index()
nu_dat.index = range(len(u_uni))
nu_dat.to_csv('experiments/debug2/input/demographics_reindexed.tsv', sep='\t', header = 0, index= False)
```
